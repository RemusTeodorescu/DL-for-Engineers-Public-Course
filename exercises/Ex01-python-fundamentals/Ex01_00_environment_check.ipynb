## What Ex_1 is about

The tools before the models: **Python, NumPy and matplotlib, used
deliberately.** There is no deep learning in this exercise set and nothing is
trained. You write and debug plain Python, learn how NumPy arrays and
broadcasting actually behave, and produce figures fit for a report — all on one
physical object, so that every array and every plot means something.

### Goals

By the end of the exercise set you can

1. write and debug Python — types, containers, functions — and read a traceback
   well enough to act on it rather than guess;
2. use NumPy arrays deliberately: dtype, shape, slicing, broadcasting, and
   replace a loop with a vectorised expression;
3. produce an engineering figure that could go into a report — labelled axes,
   units, a legend that means something;
4. recognise the arithmetic traps that arrive with you from MATLAB, in
   particular that `**` is the power operator and `^` is not;
5. check a computed result against a hand calculation instead of trusting it.

## 3 · Reproducibility

Almost every notebook in this course generates random numbers — noisy measurements here, a network's initial weights later on. Random means
*different every run*, and that is intolerable when you are trying to work out
whether a change you made improved anything.

The fix is to seed the generator. `set_seed(88)` seeds Python's `random` module
and NumPy's global generator. Once PyTorch is in use, the same call seeds it as
well. The number 88 is arbitrary; it is used throughout the course so that your output
matches the numbers quoted in the text.

Run the next cell twice. The two draws inside a single run differ from each
other — the generator advances. But the *sequence* is identical every time you
re-run the cell, because `set_seed` resets the generator to the same starting
state.

````python
set_seed(88)
first = np.random.normal(size=3)
second = np.random.normal(size=3)

print("first draw :", np.round(first, 4))
print("second draw:", np.round(second, 4))
````

**What you should see.** Two different rows of three numbers — and exactly the
same two rows every time you re-run the cell. If re-running gives you different
numbers, the `set_seed` call is not being executed; check you have not split
the cell.

A caution you will need later: seeding makes a run reproducible *on the same
machine with the same library versions*. It does not guarantee that your
numbers match a colleague's to the last decimal. When a result matters, report
what is stable across seeds, not the digits of one run.

## 4 · NumPy smoke test

Three things are checked here, and each one is a subject of notebook 02:

- an array has a **shape**, and it is printed as a tuple;
- an array has a **dtype**, decided when it is created and not afterwards;
- arithmetic between arrays of different shapes may still work, through
  **broadcasting**.

You are not expected to follow the last one yet. Notebook 02 does nothing else
for its entire length.

````python
a = np.array([1.0, 2.0, 3.0])          # 1-D, three elements
b = np.array([[10.0], [20.0]])         # 2-D, two rows and one column

describe(a, "a")
describe(b, "b")

print("\n  a + b has shape", (a + b).shape, "- that is broadcasting, and it is notebook 02")
````

**What you should see.** `a` with shape `(3,)` and dtype `float64`; `b` with
shape `(2, 1)`; and their sum with shape `(2, 3)`.

Note that `(3,)` is a tuple of length one, not a typo. Python writes a
one-element tuple with a trailing comma to distinguish it from a bracketed
expression. A shape of `(3,)` means *one axis, of length three* — which is a
genuinely different object from `(3, 1)` or `(1, 3)`, and the difference will
matter more than you would expect.

## 5 · matplotlib

The last check draws a figure, using the running example of this exercise set:
the deflection of a 2 m steel cantilever carrying 500 N/m — a loaded shelf.

The figure is drawn with the course's conventions, applied by
`engineering_axes`: both axes labelled, both labels carrying units, a light
grid, no box around the plot. Notebook 03 is about those conventions and why
they are not decoration.

````python
x = np.linspace(0.0, 2.0, 200)                 # position along the beam, m
y = beam_deflection(x)                         # deflection, mm

fig, ax = plt.subplots(figsize=(6.0, 3.4))
ax.plot(x, y, linewidth=2.0)
engineering_axes(ax, "position from fixed end x [m]", "deflection [mm]",
                 title="Cantilever under uniform load")
ax.invert_yaxis()                              # deflection is downwards
plt.show()
````

**What you should see.** A smooth curve starting at zero at the fixed end and
growing to roughly 4.0 mm at the free end, drawn downwards because the beam
deflects downwards and a figure should agree with the physical world.

If no figure appears at all, and no error was raised, your matplotlib is

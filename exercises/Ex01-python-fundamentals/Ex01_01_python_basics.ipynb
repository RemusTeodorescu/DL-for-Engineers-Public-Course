## 1 · Values and types

Python is **dynamically typed**: a variable is a name attached to a value, and
the value carries the type. You never declare a type, and the same name may
point at an integer now and a string later. That is convenient and it is also
how a certain class of bug survives to production, so it is worth being
deliberate.

The four types you will use constantly:

| type | example | note |
|---|---|---|
| `int` | `4`, `-17` | unbounded — no overflow, unlike C or MATLAB's integer types |
| `float` | `4.0`, `2.1e-3` | double precision, always |
| `str` | `"steel"` | single and double quotes are identical in meaning |
| `bool` | `True`, `False` | capitalised, and a subtype of `int` |

Three operators surprise people arriving from MATLAB:

- `/` is **always** floating-point division. `7 / 2` is `3.5` even though both
  operands are integers.
- `//` is floor division, and `%` is the remainder. `7 // 2` is `3`.
- `**` is the power operator. `^` exists but means bitwise exclusive-or, so
  `2 ^ 3` is `1` and not `8`. This one is silent and it will get you once.

`type(x)` reports a type; f-strings interpolate values into text. The `=`
inside an f-string placeholder — `f"{x = }"` — prints both the expression and
its value, which makes it the fastest debugging tool in the language.

````python
n_cycles = 250_000          # underscores are allowed in numeric literals
stress_mpa = 137.5
material = "S355 steel"
passed = stress_mpa < 200.0

print(f"{n_cycles = }  {type(n_cycles).__name__}")
print(f"{stress_mpa = }  {type(stress_mpa).__name__}")
print(f"{material = }  {type(material).__name__}")
print(f"{passed = }  {type(passed).__name__}")

print()
print("7 / 2   =", 7 / 2,   "   true division, always a float")
print("7 // 2  =", 7 // 2,  "     floor division")
print("7 % 2   =", 7 % 2,   "     remainder")
print("2 ** 3  =", 2 ** 3,  "     power")
print("2 ^ 3   =", 2 ^ 3,   "     bitwise XOR - NOT a power. Watch for this.")
````

**What you should see.** Four labelled values with their type names, then the
five arithmetic results. The last line printing `1` rather than `8` is the
point of the cell.

## 2 · Containers: list, tuple, dict

Python's built-in containers do the work that a matrix does in MATLAB, and they
are not matrices. Numerical arrays come in notebook 02; these are for
*bookkeeping* — names, settings, collections of unlike things.

**A list** is an ordered, mutable sequence, written with square brackets. It may
hold anything, including a mixture of types, and it grows with `.append`.

**A tuple** is the same thing but immutable, written with round brackets. Use
one when the length is part of the meaning — a shape, a coordinate pair, a
return value with two parts.

**A dict** maps keys to values, written with braces. It is the right container
whenever you catch yourself maintaining two parallel lists.

Two conventions to internalise:

1. **Indexing starts at zero.** `xs[0]` is the first element. `xs[-1]` is the
   last, `xs[-2]` the one before it — negative indices count from the end,
   which is genuinely convenient once you stop reading it as an error.
2. **A slice excludes its upper bound.** `xs[1:4]` gives elements 1, 2 and 3 —
   three of them, which is `4 - 1`. Every slice length is *stop minus start*.
   That property is why the convention was chosen, and it makes adjacent slices
   like `xs[:k]` and `xs[k:]` partition the list exactly.

````python
loads = [12.0, 18.5, 31.2, 44.0, 51.7]        # kN

print("whole list        ", loads)
print("first             ", loads[0])
print("last              ", loads[-1])
print("elements 1,2,3    ", loads[1:4], "  <- stop is excluded, so 3 of them")
print("every other       ", loads[::2])
print("reversed          ", loads[::-1])
print("length            ", len(loads))

section = ("IPE200", 2850.0, 1.94e-5)          # name, area mm^2, I in m^4
name, area, second_moment = section             # unpacking, one line
print()
print(f"{name}: area {area} mm^2, I = {second_moment:.2e} m^4")

steel = {"E_gpa": 210.0, "yield_mpa": 355.0, "density": 7850.0}
print()
print("E                 ", steel["E_gpa"], "GPa")
print("keys              ", list(steel.keys()))
print("missing key, safely", steel.get("poisson", "not recorded"))
````

### The aliasing trap

This one deserves its own heading because it is the Python behaviour that
surprises engineers most, and because it is silent.

Assignment in Python **never copies a container**. `b = a` makes `b` a second
name for the same list. Mutate through either name and both see the change,
because there is only one list. MATLAB's copy-on-write semantics mean the
equivalent code there does something different, so the instinct you arrive with
is wrong in a way that produces no error message.

If you want an independent list, ask for one: `b = a.copy()` or `b = list(a)`.
For nested structures, `copy.deepcopy`.

Run the next cell and read it carefully. `id(x)` returns the identity of an
object, which you can treat as its memory address; two names with the same `id`
are the same object.

````python
a = [1.0, 2.0, 3.0]
b = a                       # NOT a copy
c = a.copy()                # a genuine copy

b.append(99.0)

print("a         ", a)
print("b         ", b)
print("c         ", c)
print()
print("a is b    ", a is b, "  same object")
print("a is c    ", a is c, "  different objects")
print("a == c    ", a == c, "  ... which is why 'is' and '==' are different questions")
````

**What you should see.** `a` and `b` both ending in `99.0`, while `c` still has
three elements. `a is b` is `True`; `a == c` is `False` only because `a` grew.

Read `is` as *the same object* and `==` as *equal in value*. Use `==` for
comparisons and `is` only for `None`: the idiom `if x is None:` is the correct
way to test for absence.

## 3 · Control flow

Python marks blocks with **indentation**, not with braces or an `end` keyword.
The line that opens a block ends with a colon, and everything indented under it
belongs to the block. Four spaces per level is the near-universal convention;
mixing tabs and spaces is an error, and every sane editor is configured to
insert spaces when you press Tab.

Iterating deserves a paragraph of its own, because Python's `for` is not C's.
It does not count — it walks the elements of a sequence directly. If you find
yourself writing `for i in range(len(xs)):` and then only ever using `xs[i]`,
write `for x in xs:` instead. When you genuinely need the position as well, use
`enumerate`. When you need to walk two sequences in step, use `zip`, which stops
at the shorter one.

`range(a, b)` produces `a, a+1, ..., b-1` — the upper bound is excluded, just
as in a slice, and for the same reason: `len(range(a, b))` is `b - a`.

Finally, the **list comprehension**. `[f(x) for x in xs if cond(x)]` builds a
list in one expression. It is not merely shorter than the equivalent loop; it
is the form a Python programmer reads fastest, and you should use it whenever
the body is a single expression. When the body needs several statements, use a
loop and do not apologise for it.

````python
loads_kn = [12.0, 18.5, 31.2, 44.0, 51.7]
capacity_kn = 40.0

for i, load in enumerate(loads_kn):
    if load < 0.5 * capacity_kn:
        verdict = "comfortable"
    elif load < capacity_kn:
        verdict = "acceptable"
    else:
        verdict = "OVER CAPACITY"
    print(f"  case {i}: {load:6.1f} kN   {verdict}")

utilisation = [load / capacity_kn for load in loads_kn]
overloaded = [load for load in loads_kn if load > capacity_kn]

print()
print("utilisation:", [round(u, 3) for u in utilisation])
print("overloaded :", overloaded)

names = ["A", "B", "C", "D", "E"]
print()
for name, load in zip(names, loads_kn):
    print(f"  case {name} carries {load} kN")
````

**What you should see.** Five verdict lines, with cases 3 and 4 acceptable and
over capacity respectively; then the utilisation list; then five labelled lines
from `zip`.

## 4 · Functions

`def` introduces a function; the value comes back with `return`; a function
with no `return` returns `None`. Parameters may have defaults, and callers may
pass arguments by name in any order — which makes a call with four numeric
arguments readable instead of cryptic. Compare

```python
deflection(2.0, 500.0, 210e9, 8e-6)
deflection(length=2.0, load=500.0, E=210e9, I=8e-6)
```

The second one you can still read in six months. Prefer it whenever a call has
more than two arguments, and write your own functions so that it is possible.

The **docstring** — a string as the first statement of the function body — is
what `help(f)` prints. Write one for anything you will still be using next
week. One line is enough; say what comes back and in what units.

Returning several values is done with a tuple, usually without brackets:
`return mean, spread`. The caller unpacks it: `m, s = summarise(xs)`.

### One trap: mutable default arguments

A default argument is evaluated **once**, when the `def` statement runs — not on
each call. So a default of `[]` is a single list, shared by every call that
does not supply one, and it accumulates. The cell below demonstrates it, and
the fix is always the same: default to `None`, and build the container inside
the function.

````python
def summarise(values, label="sample"):
    """Return (mean, peak-to-peak spread) of `values`, and print a one-line report."""
    mean = sum(values) / len(values)
    spread = max(values) - min(values)
    print(f"  {label}: mean {mean:.2f}, spread {spread:.2f}")
    return mean, spread


m, s = summarise([12.0, 18.5, 31.2, 44.0, 51.7], label="loads [kN]")
print(f"  unpacked: mean = {m:.2f}, spread = {s:.2f}")


def broken_log(entry, log=[]):              # the trap
    log.append(entry)
    return log


def fixed_log(entry, log=None):             # the fix
    if log is None:
        log = []
    log.append(entry)
    return log


print()
print("shared default: ", broken_log("first"), broken_log("second"))
print("fresh each call:", fixed_log("first"), fixed_log("second"))
````

**What you should see.** The summary line, then

```
shared default:  ['first', 'second'] ['first', 'second']
fresh each call: ['first'] ['second']
```

Both calls to `broken_log` printed the same accumulating list, because there is
only one. This is not a bug in Python — it is a direct consequence of the
`def` statement being executed like any other — but it behaves like one, so
default to `None` and be done with it.

## 5 · Imports and namespaces

An import runs another file and binds a name to the resulting module. Three
forms, all in use:

```python
import numpy as np             # the module, under a short alias
from math import pi, sqrt      # two names, lifted into this namespace
from Ex_1_core import *        # everything the module exports
```

The first form is the default. It keeps the origin of every name visible:
`np.linspace` is unmistakably NumPy's. The second is fine for a handful of
names you use constantly. The third is discouraged in real code — it fills your
namespace with names of unknown origin, and two modules doing it will silently
overwrite each other — but it is what the setup cell of every notebook in this
course uses, because there is one module and the alternative is prefixing
every call in every exercise.

**Where does Python look?** In order: the directory the notebook is running
from, then the installed packages. That first entry is the whole reason the
exercise instructions insist that the module sits next to the notebook, and it
is the reason `FileNotFoundError` on Colab is fixed by re-uploading rather than
by re-installing anything.

`module.__file__` shows which file was actually loaded, which settles the
question when you suspect you are importing the wrong copy of something. That
suspicion is correct more often than you would think.

````python
import math
import Ex_1_core

print("math.pi          ", math.pi)
print("math.sqrt(2)     ", math.sqrt(2))
print()
print("loaded from      ", Ex_1_core.__file__)
print()
print("first few entries on the search path:")
import sys
for entry in sys.path[:3]:
    print("   ", entry if entry else "(the current working directory)")
````

## 6 · Reading a traceback

This is the most valuable half-page in the notebook.

When Python raises an exception it prints a **traceback**: the chain of calls
that led to the failure, oldest first, with the error itself at the very
bottom. Most people glance at it, see a wall of text, and go back to staring at
their code. That is exactly backwards. The traceback already contains the
answer; it just has to be read in the right order.

**Read it from the bottom up.**

1. The **last line** is the exception type and its message: *what* went wrong.
   `IndexError: list index out of range`. `TypeError: unsupported operand
   type(s)`. Read this first and read it literally.
2. The **frame directly above it** — the last `File ..., line N, in f` block —
   is *where* it went wrong, with the offending source line shown.
3. The frames **above that** are *how you got there*: the call chain from your
   cell down into the library. Scan upward until you find the last frame that
   is in code you wrote. That is nearly always the line you must change, even
   when the exception was raised deep inside NumPy.

Run the next cell. `traceback_demo()` raises from three frames down, on
purpose. Practise on it: name the exception type, name the function that
raised, and name the line in `traceback_demo` that started the chain — before
reading on.

````python
traceback_demo()
````

**What you should see.** An `IndexError: list index out of range`, raised in
`inner`, called from `middle`, called from `outer`, called from the cell you
just ran. The list has three elements and the index requested was 5.

This cell is *meant* to raise, so **Run All** will stop here. Carry on with
Shift+Enter.

Note what the traceback did *not* tell you: that the index was 5, or that the
list had three entries. It told you where to look, and you found those two
facts by looking. That division of labour is the normal case. When the message
alone is not enough, print the shapes and lengths of everything on the line
above the failure — which in NumPy code means printing `.shape`, every time,
and is the technique notebook 02 is built around.

## 7 · A short list of things that will catch you

Collected here so you can come back to it. Every one of these has cost someone
in this course at least an hour.

| | |
|---|---|
| **Indexing starts at 0** | `xs[0]` is the first element, `xs[-1]` the last. |
| **Slices exclude the stop** | `xs[1:4]` has three elements; `range(0, 5)` stops at 4. |
| **`^` is not a power** | `2 ** 3` is 8; `2 ^ 3` is 1, silently. |
| **`=` does not copy** | `b = a` aliases a list. Use `a.copy()`. |
| **Mutable defaults persist** | Default to `None` and build inside. |
| **Indentation is syntax** | Four spaces. Never mix tabs and spaces. |
| **Integer division** | `/` is float division even for two ints; `//` floors. |
| **Modules import once** | Edit a `.py` and you must restart the kernel. |
| **Cells run in the order you press them** | Not top to bottom. Restart and Run All when confused. |
| **`is` is not `==`** | `is` asks about identity. Use it only with `None`. |
